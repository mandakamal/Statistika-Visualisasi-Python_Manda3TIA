## 📋 Tabel Frekuensi Grade

```python
tabel_frekuensi = (
    data["Grade"]
    .value_counts()
    .reindex(["A", "AB", "B", "BC", "C", "F"], fill_value=0)
)

tabel_frekuensi
```

```python
tabel_persentase = (
    data["Grade"]
    .value_counts(normalize=True)
    .reindex(["A", "AB", "B", "BC", "C", "F"], fill_value=0)
    * 100
)

tabel_grade = pd.DataFrame({
    "Grade": tabel_frekuensi.index,
    "Frekuensi": tabel_frekuensi.values,
    "Persentase (%)": tabel_persentase.round(2).values
})

tabel_grade
```

## 📈 Line Chart — Distribusi Frekuensi Grade

```python
ax = tabel_frekuensi.plot(
    kind="line",
    marker="o",
    figsize=(9, 5),
    title="Distribusi Frekuensi Grade Mahasiswa"
)

ax.set_xlabel("Grade")
ax.set_ylabel("Frekuensi")
```

## 📊 Bar Chart — Jumlah Mahasiswa Berdasarkan Grade

```python
ax = tabel_frekuensi.plot(
    kind="bar",
    figsize=(9, 5),
    title="Jumlah Mahasiswa Berdasarkan Grade"
)

ax.set_xlabel("Grade")
ax.set_ylabel("Frekuensi")
```

## 🥧 Pie Chart — Persentase Grade Mahasiswa

```python
ax = tabel_frekuensi.plot(
    kind="pie",
    figsize=(7, 7),
    autopct="%1.1f%%",
    title="Persentase Grade Mahasiswa"
)

ax.set_ylabel("")
```

# 📊 Statistika Deskriptif Nilai Akhir

```python
nilai = pd.to_numeric(
    data["Nilai Akhir"],
    errors="coerce"
).dropna()

nilai
```

```python
mean = nilai.mean()

print("Mean:", round(mean, 2))
```

```python
median = nilai.median()

print("Median:", round(median, 2))
```

```python
nilai.mode()[0]
```

```python
modus = nilai.mode()

print("Modus:")
print(modus)
```

```python
minimum = nilai.min()
maximum = nilai.max()

print("Minimum:", minimum)
print("Maximum:", maximum)
```

```python
q1 = nilai.quantile(0.25)
q2 = nilai.quantile(0.50)
q3 = nilai.quantile(0.75)

print("Q1:", round(q1, 2))
print("Q2:", round(q2, 2))
print("Q3:", round(q3, 2))
```

```python
range_nilai = maximum - minimum

print("Range:", round(range_nilai, 2))
```

```python
iqr = q3 - q1

print("IQR:", round(iqr, 2))
```

```python
variance = nilai.var()

print("Variance:", round(variance, 2))
```

```python
std_dev = nilai.std()

print("Standard Deviation:", round(std_dev, 2))
```

```python
standard_error = std_dev / (len(nilai) ** 0.5)

print("Standard Error:", round(standard_error, 2))
```

```python
skewness = nilai.skew()

print("Skewness:", round(skewness, 3))
```

```python
kurtosis = nilai.kurt()

print("Kurtosis:", round(kurtosis, 3))
```

```python
statistik = pd.DataFrame({
    "Statistik": [
        "Jumlah Mahasiswa",
        "Mean",
        "Median",
        "Modus",
        "Minimum",
        "Q1 (25%)",
        "Q2 (50%)",
        "Q3 (75%)",
        "Maximum",
        "Range",
        "IQR",
        "Variance",
        "Standard Deviation",
        "Standard Error",
        "Skewness",
        "Kurtosis"
    ],

    "Nilai": [
        len(nilai),
        mean,
        median,
        ", ".join([str(round(x, 2)) for x in modus]),
        minimum,
        q1,
        q2,
        q3,
        maximum,
        range_nilai,
        iqr,
        variance,
        std_dev,
        standard_error,
        skewness,
        kurtosis
    ]
})

statistik
```

## 🔎 Interpretasi Hasil Statistik

```python
print("=== INTERPRETASI HASIL ===")

print(f"Rata-rata nilai mahasiswa adalah {mean:.2f}.")
print(f"Median nilai mahasiswa adalah {median:.2f}.")
print(f"Nilai minimum adalah {minimum:.2f}, sedangkan nilai maksimum adalah {maximum:.2f}.")
print(f"Sebaran 50% data bagian tengah berada antara Q1 = {q1:.2f} dan Q3 = {q3:.2f}.")
print(f"Nilai IQR sebesar {iqr:.2f}.")
print(f"Standard deviation sebesar {std_dev:.2f}.")
print(f"Skewness sebesar {skewness:.3f}.")
print(f"Kurtosis sebesar {kurtosis:.3f}.")
```
